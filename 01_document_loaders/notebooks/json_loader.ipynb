```python
from langchain_community.document_loaders.json_loader import JSONLoader
from pathlib import Path
from pprint import pprint as pp
```

```python
file_path = Path("../knowledge-source/apparels.json")
print(file_path.exists())
```

```python
def metadata_func(record: dict, metadata: dict) -> dict:
    # Extract metadata from the record and add it to the metadata dictionary
    metadata["ProductId"] = record.get("ProductId")
    metadata["Name"] = record.get("Name")
    metadata["Category"] = record.get("Category")
    metadata["Price"] = record.get("Price")
    return metadata
```

```python
json_loader = JSONLoader(file_path=file_path.as_posix(),
                         jq_schema = ".products[]",
                        content_key="Description",
                        metadata_func=metadata_func
                         )
```

```python
json_documents = json_loader.load()
print(len(json_documents))
pp(json_documents[0:5])  #
```
