```python
from langchain_community.document_loaders.pdf import PyPDFLoader
from pathlib import Path
from pprint import pp  
```

```python
# Define the path to the text file
file_path = Path("../knowledge-source/attention_is_all_you_need.pdf")
print(file_path)  # Print the file path
print(file_path.exists())  # Check if the file exists
```

```python
# Create the loader
pypdf_loader = PyPDFLoader(file_path=file_path.as_posix(), mode="page")
```

```python
# Load the documents
documents = pypdf_loader.load()
```

```python
# Check the contents
print(f"Number of documents loaded: {len(documents)}")
print(f"First document content: {documents[0].page_content[:500]}")  # Print the first 500 characters of the first document
print(f"Metadata of the first document: {documents[0].metadata}")
```

```python
# Extract image from PDF
from langchain_community.document_loaders.parsers import TesseractBlobParser
```

```python
pypdf_image_loader = PyPDFLoader(file_path=file_path.as_posix(),
                                mode="page",
                                images_parser=TesseractBlobParser(),
                                images_inner_format="html-img"
                                )
```

```python
document_with_images = pypdf_image_loader.load()
```

```python
print(len(document_with_images))
print(f"First document content with images: {document_with_images[2].page_content[-500:]}")
pp(f"First document metadata with images: {document_with_images[2].metadata}")  # Print the metadata of the first document with images
```

```python
print(document_with_images[5].page_content[:500])  # Print the first 500 characters of the first document with images
```
