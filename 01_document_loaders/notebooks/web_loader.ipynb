```python
from langchain_community.document_loaders import WebBaseLoader
from pprint import pp
```

```python
url = "https://shop.hillspet.com"
web_loader = WebBaseLoader(url)
documents = web_loader.load()
```

```python
print(len(documents))
pp(documents[0].page_content)
```

# Recursive loaders

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
base_url = "https://docs.python.org/3/tutorial/"
recursive_loader = RecursiveUrlLoader(url=base_url, max_depth=2)
```

```python
recursive_documents = recursive_loader.load()
```

```python
print(len(recursive_documents))
pp(recursive_documents[0].page_content)
```

```python
# Lazy load
documents_lazy = web_loader.lazy_load()
```

```python
counter = 0
for doc in documents_lazy:
    counter += 1
    print(f"Document {counter}:")
    pp(doc.page_content)
    print("\n---\n")
```
