```python
from langchain_community.document_loaders.csv_loader import CSVLoader
from pathlib import Path
from pprint import pprint as pp
```

```python
file_path = Path("../knowledge-source/organizations.csv")
csv_loader = CSVLoader(file_path=file_path.as_posix(),
                       source_column="Industry",
                       encoding="utf-8",
                       metadata_columns=["Index","Organization Id","Name","Website","Country","Founded","Number of employees"],
                       content_columns="Description")
print(file_path.exists())
```

```python
csv_documents = csv_loader.load()
print(len(csv_documents))
```
